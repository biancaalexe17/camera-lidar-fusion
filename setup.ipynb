```python
import subprocess
from pathlib import Path

pp_env = Path("/content/pp_env")

if pp_env.exists():
    print("Existing environment found:", pp_env)
    subprocess.run(
        [str(pp_env / "bin/python"), "--version"],
        check=False
    )
else:
    subprocess.run([
        "uv", "venv", str(pp_env),
        "--python",
        "/root/.local/share/uv/python/cpython-3.11-linux-x86_64-gnu/bin/python3.11"
    ], check=True)

subprocess.run([str(pp_env / "bin/python"), "--version"], check=True)
```
